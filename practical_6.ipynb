Matrix Chain Multiplication

```python
import time

# Matrix Chain Multiplication
def matrix_chain(p):
    n = len(p) - 1

    # DP table
    dp = [[0] * n for _ in range(n)]

    # Chain length
    for length in range(2, n + 1):

        for i in range(n - length + 1):
            j = i + length - 1

            dp[i][j] = float('inf')

            # Find minimum cost
            for k in range(i, j):

                cost = (dp[i][k] +
                        dp[k + 1][j] +
                        p[i] * p[k + 1] * p[j + 1])

                dp[i][j] = min(dp[i][j], cost)

    return dp[0][n - 1]


# ---------------- USER INPUT ----------------

n = int(input("Enter number of matrices: "))

p = []

print("\nEnter matrix dimensions:")

rows = int(input("Enter rows of Matrix 1: "))
cols = int(input("Enter columns of Matrix 1: "))

p.append(rows)
p.append(cols)

for i in range(2, n + 1):

    rows = int(input(f"Enter rows of Matrix {i}: "))

    # Check matrix compatibility
    if rows != p[-1]:
        print("Error: Matrices are not compatible!")
        exit()

    cols = int(input(f"Enter columns of Matrix {i}: "))
    p.append(cols)


# ---------------- EXECUTION TIME ----------------

start_time = time.perf_counter()

result = matrix_chain(p)

end_time = time.perf_counter()

execution_time = end_time - start_time


# ---------------- OUTPUT ----------------

print("\n----- RESULT -----")
print("Minimum number of scalar multiplications:", result)

print("Execution Time:", execution_time, "seconds")

print("\n----- COMPLEXITY -----")
print("Time Complexity: O(n^3)")
print("Space Complexity: O(n^2)")
```

The time complexity is **O(n³)** and the space complexity is **O(n²)**. This approach efficiently determines the optimal order of matrix multiplication.
