```python

import time

n = int(input("Enter number of vertices: "))

print("Enter adjacency matrix (use 0 for no edge):")
graph = []

for i in range(n):
    row = list(map(int, input(f"Enter row {i + 1}: ").split()))
    graph.append(row)

start = time.time()

# Maximum Floyd-Warshall Algorithm
for k in range(n):
    for i in range(n):
        for j in range(n):
            graph[i][j] = max(
                graph[i][j],
                min(graph[i][k], graph[k][j])
            )

end = time.time()

print("\nMaximum Capacity Matrix:")
for row in graph:
    print(*row)

print("\nTime Complexity: O(n^3)")
print("Execution Time:", end - start, "seconds")
```

Conclusion:
The Maximum Floyd–Warshall algorithm finds the maximum capacity path between every pair of vertices in a graph using dynamic programming. It has a time complexity of \(O(n^3)\) and a space complexity of \(O(n^2)\). The execution time depends on the input size and computer performance
