```python
import time

# Input number of vertices
n = int(input("Enter number of vertices: "))

# Input number of edges
e = int(input("Enter number of edges: "))

# Create graph
graph = [[0 for _ in range(n)] for _ in range(n)]

print("Enter edges: source destination weight")

for i in range(e):
    u, v, w = map(int, input(f"Edge {i + 1}: ").split())
    graph[u][v] = w
    graph[v][u] = w


# Start time
start = time.perf_counter()

# Prim's Algorithm
visited = [False] * n
min_edge = [float('inf')] * n
parent = [-1] * n

min_edge[0] = 0
total_cost = 0

for _ in range(n):

    # Find minimum edge
    u = -1
    minimum = float('inf')

    for i in range(n):
        if not visited[i] and min_edge[i] < minimum:
            minimum = min_edge[i]
            u = i

    # Check for disconnected graph
    if u == -1:
        print("Graph is disconnected. MST cannot be formed.")
        exit()

    visited[u] = True
    total_cost += min_edge[u]

    # Update adjacent vertices
    for v in range(n):
        if graph[u][v] != 0 and not visited[v]:
            if graph[u][v] < min_edge[v]:
                min_edge[v] = graph[u][v]
                parent[v] = u


# End time
end = time.perf_counter()

# Display MST
print("\nMinimum Spanning Tree:")

for i in range(1, n):
    print(parent[i], "--", i, "weight =", graph[parent[i]][i])

print("\nTotal cost of MST:", total_cost)

# Execution time
execution_time = end - start
print("Execution time:", execution_time, "seconds")

# Time complexity
print("Time Complexity: O(V^2)")
```

Conclusion:
Prim’s Algorithm successfully finds the Minimum Spanning Tree (MST) of a connected weighted graph. It selects the minimum-weight edge at each step and connects all vertices with minimum total cost. The time complexity of this implementation is O(V²), and the execution time is measured using Python’s
