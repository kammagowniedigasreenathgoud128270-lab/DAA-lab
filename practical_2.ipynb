```python
# Linear Search

n = int(input("Enter the number of elements: "))

arr = []
print("Enter the elements:")
for i in range(n):
    arr.append(int(input()))

target = int(input("Enter the element to search: "))

index = -1

for i in range(n):
    if arr[i] == target:
        index = i
        break

if index != -1:
    print("Element found at index:", index)
else:
    print("Element not found")

print("\nTime Complexity:")
print("Best Case: O(1)")
print("Average Case: O(n)")
print("Worst Case: O(n)")

print("Space Complexity: O(1)")
```

```python
# Binary Search

n = int(input("Enter the number of elements: "))

arr = []
print("Enter the sorted elements:")
for i in range(n):
    arr.append(int(input()))

target = int(input("Enter the element to search: "))

low = 0
high = n - 1
index = -1

while low <= high:
    mid = (low + high) // 2

    if arr[mid] == target:
        index = mid
        break
    elif arr[mid] < target:
        low = mid + 1
    else:
        high = mid - 1

if index != -1:
    print("Element found at index:", index)
else:
    print("Element not found")

print("\nTime Complexity:")
print("Best Case: O(1)")
print("Average Case: O(log n)")
print("Worst Case: O(log n)")

print("Space Complexity: O(1)")
```
