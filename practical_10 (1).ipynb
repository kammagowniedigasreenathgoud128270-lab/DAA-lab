```python
import time

# Find function
def find(parent, vertex):
    if parent[vertex] != vertex:
        parent[vertex] = find(parent, parent[vertex])
    return parent[vertex]


# Union function
def union(parent, rank, u, v):
    root_u = find(parent, u)
    root_v = find(parent, v)

    if root_u != root_v:
        if rank[root_u] < rank[root_v]:
            parent[root_u] = root_v
        elif rank[root_u] > rank[root_v]:
            parent[root_v] = root_u
        else:
            parent[root_v] = root_u
            rank[root_u] += 1


# User input
vertices = int(input("Enter number of vertices: "))
edges = int(input("Enter number of edges: "))

graph = []

print("\nEnter edges as: source destination weight")
print("Example: 0 1 10")

for i in range(edges):
    u, v, w = map(int, input(f"Edge {i + 1}: ").split())
    graph.append((w, u, v))


# Start execution time
start_time = time.perf_counter()

# Sort edges by weight
graph.sort()

# Create parent and rank arrays
parent = list(range(vertices))
rank = [0] * vertices

mst = []
total_weight = 0

# Kruskal's Algorithm
for weight, u, v in graph:

    root_u = find(parent, u)
    root_v = find(parent, v)

    # Add edge if it does not create a cycle
    if root_u != root_v:
        mst.append((u, v, weight))
        total_weight += weight

        union(parent, rank, root_u, root_v)

        # MST contains V-1 edges
        if len(mst) == vertices - 1:
            break


# End execution time
end_time = time.perf_counter()


# Display result
if len(mst) != vertices - 1:
    print("\nGraph is disconnected. MST cannot be formed.")
else:
    print("\nMinimum Spanning Tree:")

    for u, v, weight in mst:
        print(f"{u} -- {v} = {weight}")

    print("\nTotal weight of MST:", total_weight)

    execution_time = end_time - start_time

    print("Execution time: {:.8f} seconds".format(execution_time))
    print("Time Complexity: O(E log E)")
```

Conclusion
Kruskal’s Algorithm successfully finds the Minimum Spanning Tree (MST) by selecting the edges with the smallest weights without forming cycles. The algorithm connects all vertices with minimum total cost and has a time complexity of O(E log E).
